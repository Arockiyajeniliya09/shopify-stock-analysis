```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python
df = pd.read_csv("/content/shopify_stock.csv")
```

```python
df['date'] = pd.to_datetime(df['date'], utc=True)
df = df.sort_values('date').reset_index(drop=True)
```

```python
print("--- Dataset Overview ---")
print(df.info())
print("\n--- Summary Statistics ---")
print(df.describe().T)
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 10))
```

```python
plt.subplot(2, 1, 1)
plt.plot(df['date'], df['adj_close'], color='#008060', linewidth=1.5, label='Adjusted Close')
plt.title('Shopify Stock Price History (Adj Close)', fontsize=14, fontweight='bold')
plt.ylabel('Price (USD)')
plt.legend(loc='upper left')
```

```python
plt.subplot(2, 1, 2)
plt.bar(df['date'], df['volume'], color='#5c6ac4', alpha=0.6, width=1)
plt.title('Daily Trading Volume', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Volume')

plt.tight_layout()
plt.show()
```

```python
df['MA50'] = df['adj_close'].rolling(50).mean()
df['MA200'] = df['adj_close'].rolling(200).mean()

plt.figure(figsize=(14, 6))
plt.plot(df['date'], df['adj_close'], alpha=0.4, label='Adj Close', color='gray')
plt.plot(df['date'], df['MA50'], label='50-Day Moving Average', color='blue', linewidth=1.5)
plt.plot(df['date'], df['MA200'], label='200-Day Moving Average', color='orange', linewidth=1.5)
plt.title('Shopify Stock Price with 50-Day & 200-Day Moving Averages', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```
